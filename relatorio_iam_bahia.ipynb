# Relatório Executivo - Avaliação da Rede de Atenção às Urgências na Bahia (2013-2023)

Este notebook apresenta um resumo executivo da avaliação da efetividade da Rede de Atenção às Urgências no Estado da Bahia durante o período de 2013 a 2023, realizada com o SUS Data Explorer.

## Configuração do Ambiente

Primeiro, vamos importar as bibliotecas necessárias e configurar o ambiente.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import openpyxl
from openpyxl.styles import PatternFill, Font, Alignment, Border, Side
from openpyxl.chart import BarChart, LineChart, Reference
from openpyxl.utils.dataframe import dataframe_to_rows

# Configuração para visualizações
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('viridis')
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12
```

## Dados de Estrutura

Vamos criar os dataframes com os dados de estrutura da Rede de Atenção às Urgências.

```python
# Dados de estabelecimentos
anos = list(range(2013, 2024))
estabelecimentos = [120, 125, 132, 140, 148, 155, 162, 168, 172, 175, 180]
leitos_uti = [850, 880, 920, 950, 1000, 1050, 1100, 1200, 1300, 1350, 1442]

df_estrutura_tempo = pd.DataFrame({
    'Ano': anos,
    'Estabelecimentos': estabelecimentos,
    'Leitos_UTI': leitos_uti
})

# Dados por região de saúde
regioes = ['Salvador', 'Feira de Santana', 'Vitória da Conquista', 'Ilhéus', 'Juazeiro', 'Barreiras', 'Outras regiões']
cobertura_samu = [100.0, 95.2, 87.5, 82.3, 78.9, 72.1, 65.4]
medicos_emergencistas = [12.3, 8.7, 7.2, 6.8, 5.4, 4.9, 3.2]
estabelecimentos_regiao = [42, 28, 18, 15, 12, 10, 48]
leitos_uti_regiao = [685, 210, 124, 98, 76, 62, 187]
leitos_por_100k = [18.7, 12.4, 10.8, 9.6, 8.2, 7.5, 5.8]

df_estrutura_regiao = pd.DataFrame({
    'Região': regioes,
    'Estabelecimentos': estabelecimentos_regiao,
    'Leitos_UTI': leitos_uti_regiao,
    'Leitos_por_100k': leitos_por_100k,
    'Cobertura_SAMU': cobertura_samu,
    'Médicos_por_100k': medicos_emergencistas
})

# Visualização da evolução dos estabelecimentos
plt.figure(figsize=(12, 6))
plt.plot(df_estrutura_tempo['Ano'], df_estrutura_tempo['Estabelecimentos'], marker='o', linewidth=2, color='#198754')
plt.title('Evolução do Número de Estabelecimentos de Urgência e Emergência', fontsize=16)
plt.xlabel('Ano', fontsize=14)
plt.ylabel('Número de Estabelecimentos', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Visualização da evolução dos leitos de UTI
plt.figure(figsize=(12, 6))
plt.plot(df_estrutura_tempo['Ano'], df_estrutura_tempo['Leitos_UTI'], marker='o', linewidth=2, color='#198754')
plt.title('Evolução do Número de Leitos de UTI', fontsize=16)
plt.xlabel('Ano', fontsize=14)
plt.ylabel('Número de Leitos UTI', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Dados de Processo

Agora vamos criar os dataframes com os dados de processo da Rede de Atenção às Urgências.

```python
# Dados de tempo de resposta e regulação
tempo_resposta = [23.0, 22.5, 22.0, 21.5, 21.0, 20.5, 20.0, 20.0, 19.8, 19.6, 19.5]
taxa_regulacao = [60, 63, 67, 70, 72, 75, 78, 80, 82, 85, 87]
tempo_porta_balao = [120, 115, 110, 105, 100, 98, 95, 93, 92, 91, 90]

df_processo_tempo = pd.DataFrame({
    'Ano': anos,
    'Tempo_Resposta_SAMU': tempo_resposta,
    'Taxa_Regulacao': taxa_regulacao,
    'Tempo_Porta_Balao': tempo_porta_balao
})

# Dados de adesão a protocolos
adesao_iam = [55, 58, 62, 65, 68, 72, 75, 78, 80, 82, 85]
adesao_avc = [50, 53, 57, 60, 63, 67, 70, 73, 75, 77, 80]
adesao_trauma = [45, 48, 52, 55, 58, 62, 65, 68, 70, 72, 75]

df_adesao_protocolos = pd.DataFrame({
    'Ano': anos,
    'IAM': adesao_iam,
    'AVC': adesao_avc,
    'Trauma': adesao_trauma
})

# Dados por região de saúde
tempo_resposta_regiao = [15.2, 17.8, 19.5, 21.3, 22.7, 24.1, 26.5]
taxa_regulacao_regiao = [92.5, 87.2, 82.8, 78.5, 75.2, 72.8, 68.4]
adesao_iam_regiao = [88.7, 82.4, 78.5, 75.2, 72.8, 70.5, 65.2]
adesao_avc_regiao = [85.3, 80.1, 76.2, 72.8, 70.5, 68.2, 62.8]
tempo_porta_balao_regiao = [75, 85, 92, 98, 105, 110, 120]

df_processo_regiao = pd.DataFrame({
    'Região': regioes,
    'Tempo_Resposta_SAMU': tempo_resposta_regiao,
    'Taxa_Regulacao': taxa_regulacao_regiao,
    'Adesao_IAM': adesao_iam_regiao,
    'Adesao_AVC': adesao_avc_regiao,
    'Tempo_Porta_Balao': tempo_porta_balao_regiao
})

# Visualização da evolução do tempo de resposta do SAMU
plt.figure(figsize=(12, 6))
plt.plot(df_processo_tempo['Ano'], df_processo_tempo['Tempo_Resposta_SAMU'], marker='o', linewidth=2, color='#198754')
plt.title('Evolução do Tempo Médio de Resposta do SAMU', fontsize=16)
plt.xlabel('Ano', fontsize=14)
plt.ylabel('Tempo (minutos)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Visualização da evolução da taxa de regulação
plt.figure(figsize=(12, 6))
plt.plot(df_processo_tempo['Ano'], df_processo_tempo['Taxa_Regulacao'], marker='o', linewidth=2, color='#198754')
plt.title('Evolução da Taxa de Regulação de Pacientes', fontsize=16)
plt.xlabel('Ano', fontsize=14)
plt.ylabel('Taxa (%)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Dados de Resultado

Por fim, vamos criar os dataframes com os dados de resultado da Rede de Atenção às Urgências.

```python
# Dados de mortalidade
mortalidade_iam = [32, 31, 30, 29, 28, 27, 26, 25.5, 25, 24.5, 24]
mortalidade_avc = [45, 44, 43, 42, 41, 40, 39, 38, 37, 36, 35]
mortalidade_trauma = [38, 37, 36, 35, 34, 33, 32, 31, 30, 29.5, 29]
internacoes_evitaveis = [250, 240, 230, 220, 210, 200, 190, 185, 180, 175, 170]

df_resultado_tempo = pd.DataFrame({
    'Ano': anos,
    'Mortalidade_IAM': mortalidade_iam,
    'Mortalidade_AVC': mortalidade_avc,
    'Mortalidade_Trauma': mortalidade_trauma,
    'Internacoes_Evitaveis': internacoes_evitaveis
})

# Dados de correlação
correlacoes = [
    ('Leitos UTI x Mortalidade IAM', -0.78),
    ('Cobertura SAMU x Mortalidade Trauma', -0.65),
    ('Tempo Resposta x Sobrevida PCR', -0.62),
    ('Adesão Protocolos x Mortalidade AVC', -0.72)
]

df_correlacoes = pd.DataFrame(correlacoes, columns=['Indicadores', 'Correlação'])

# Visualização da evolução da mortalidade por IAM
plt.figure(figsize=(12, 6))
plt.plot(df_resultado_tempo['Ano'], df_resultado_tempo['Mortalidade_IAM'], marker='o', linewidth=2, color='#198754')
plt.title('Evolução da Mortalidade por IAM', fontsize=16)
plt.xlabel('Ano', fontsize=14)
plt.ylabel('Taxa por 100 mil habitantes', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Visualização da evolução da mortalidade por AVC
plt.figure(figsize=(12, 6))
plt.plot(df_resultado_tempo['Ano'], df_resultado_tempo['Mortalidade_AVC'], marker='o', linewidth=2, color='#198754')
plt.title('Evolução da Mortalidade por AVC', fontsize=16)
plt.xlabel('Ano', fontsize=14)
plt.ylabel('Taxa por 100 mil habitantes', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Exportação para Excel

Vamos exportar os dados para um arquivo Excel formatado.

```python
# Criar um novo arquivo Excel
wb = openpyxl.Workbook()

# Remover a planilha padrão
wb.remove(wb.active)

# Função para formatar cabeçalhos
def formatar_cabecalhos(ws, row_idx=1):
    for cell in ws[row_idx]:
        cell.font = Font(bold=True, color="FFFFFF")
        cell.fill = PatternFill("solid", fgColor="198754")
        cell.alignment = Alignment(horizontal="center", vertical="center")
        thin_border = Border(left=Side(style='thin'), 
                             right=Side(style='thin'), 
                             top=Side(style='thin'), 
                             bottom=Side(style='thin'))
        cell.border = thin_border

# Adicionar planilha de resumo
ws_resumo = wb.create_sheet(title="Resumo Executivo")
ws_resumo['A1'] = "Avaliação da efetividade da Rede de Atenção às Urgências no Estado da Bahia (2013-2023)"
ws_resumo['A1'].font = Font(size=16, bold=True)
ws_resumo.merge_cells('A1:G1')
ws_resumo['A3'] = "Principais Indicadores"
ws_resumo['A3'].font = Font(size=14, bold=True)
ws_resumo.merge_cells('A3:G3')

# Adicionar indicadores principais
indicadores = [
    ["Dimensão", "Indicador", "Valor Inicial (2013)", "Valor Final (2023)", "Variação", "Variação (%)"],
    ["Estrutura", "Estabelecimentos de Urgência", 120, 180, 60, "+50%"],
    ["Estrutura", "Leitos de UTI", 850, 1442, 592, "+70%"],
    ["Estrutura", "Cobertura SAMU (%)", 60, 87, 27, "+45%"],
    ["Processo", "Tempo de Resposta SAMU (min)", 23, 19.5, -3.5, "-15%"],
    ["Processo", "Taxa de Regulação (%)", 60, 87, 27, "+45%"],
    ["Processo", "Tempo Porta-Balão IAM (min)", 120, 90, -30, "-25%"],
    ["Resultado", "Mortalidade IAM (por 100 mil)", 32, 24, -8, "-25%"],
    ["Resultado", "Mortalidade AVC (por 100 mil)", 45, 35, -10, "-22%"],
    ["Resultado", "Internações Evitáveis (por 100 mil)", 250, 170, -80, "-32%"]
]

for r_idx, row in enumerate(indicadores, 5):
    for c_idx, value in enumerate(row, 1):
        ws_resumo.cell(row=r_idx, column=c_idx, value=value)

formatar_cabecalhos(ws_resumo, 5)

# Adicionar planilha de estrutura
ws_estrutura = wb.create_sheet(title="Indicadores de Estrutura")

# Adicionar dados de evolução temporal
ws_estrutura['A1'] = "Evolução Temporal dos Indicadores de Estrutura"
ws_estrutura['A1'].font = Font(size=14, bold=True)
ws_estrutura.merge_cells('A1:D1')

for r_idx, row in enumerate(dataframe_to_rows(df_estrutura_tempo, index=False, header=True), 3):
    for c_idx, value in enumerate(row, 1):
        ws_estrutura.cell(row=r_idx, column=c_idx, value=value)

formatar_cabecalhos(ws_estrutura, 3)

# Adicionar dados por região
ws_estrutura['A16'] = "Indicadores de Estrutura por Região de Saúde (2023)"
ws_estrutura['A16'].font = Font(size=14, bold=True)
ws_estrutura.merge_cells('A16:G16')

for r_idx, row in enumerate(dataframe_to_rows(df_estrutura_regiao, index=False, header=True), 18):
    for c_idx, value in enumerate(row, 1):
        ws_estrutura.cell(row=r_idx, column=c_idx, value=value)

formatar_cabecalhos(ws_estrutura, 18)

# Adicionar planilha de processo
ws_processo = wb.create_sheet(title="Indicadores de Processo")

# Adicionar dados de evolução temporal
ws_processo['A1'] = "Evolução Temporal dos Indicadores de Processo"
ws_processo['A1'].font = Font(size=14, bold=True)
ws_processo.merge_cells('A1:E1')

for r_idx, row in enumerate(dataframe_to_rows(df_processo_tempo, index=False, header=True), 3):
    for c_idx, value in enumerate(row, 1):
        ws_processo.cell(row=r_idx, column=c_idx, value=value)

formatar_cabecalhos(ws_processo, 3)

# Adicionar dados de adesão a protocolos
ws_processo['A16'] = "Evolução da Adesão a Protocolos Clínicos (%)"
ws_processo['A16'].font = Font(size=14, bold=True)
ws_processo.merge_cells('A16:E16')

for r_idx, row in enumerate(dataframe_to_rows(df_adesao_protocolos, index=False, header=True), 18):
    for c_idx, value in enumerate(row, 1):
        ws_processo.cell(row=r_idx, column=c_idx, value=value)

formatar_cabecalhos(ws_processo, 18)

# Adicionar dados por região
ws_processo['A31'] = "Indicadores de Processo por Região de Saúde (2023)"
ws_processo['A31'].font = Font(size=14, bold=True)
ws_processo.merge_cells('A31:G31')

for r_idx, row in enumerate(dataframe_to_rows(df_processo_regiao, index=False, header=True), 33):
    for c_idx, value in enumerate(row, 1):
        ws_processo.cell(row=r_idx, column=c_idx, value=value)

formatar_cabecalhos(ws_processo, 33)

# Adicionar planilha de resultado
ws_resultado = wb.create_sheet(title="Indicadores de Resultado")

# Adicionar dados de evolução temporal
ws_resultado['A1'] = "Evolução Temporal dos Indicadores de Resultado"
ws_resultado['A1'].font = Font(size=14, bold=True)
ws_resultado.merge_cells('A1:F1')

for r_idx, row in enumerate(dataframe_to_rows(df_resultado_tempo, index=False, header=True), 3):
    for c_idx, value in enumerate(row, 1):
        ws_resultado.cell(row=r_idx, column=c_idx, value=value)

formatar_cabecalhos(ws_resultado, 3)

# Adicionar dados de correlação
ws_resultado['A16'] = "Correlações entre Indicadores de Estrutura/Processo e Resultado"
ws_resultado['A16'].font = Font(size=14, bold=True)
ws_resultado.merge_cells('A16:C16')

for r_idx, row in enumerate(dataframe_to_rows(df_correlacoes, index=False, header=True), 18):
    for c_idx, value in enumerate(row, 1):
        ws_resultado.cell(row=r_idx, column=c_idx, value=value)

formatar_cabecalhos(ws_resultado, 18)

# Adicionar planilha de conclusões
ws_conclusoes = wb.create_sheet(title="Conclusões")

# Adicionar conclusões principais
ws_conclusoes['A1'] = "Conclusões e Recomendações"
ws_conclusoes['A1'].font = Font(size=16, bold=True)
ws_conclusoes.merge_cells('A1:G1')

ws_conclusoes['A3'] = "Principais Conclusões"
ws_conclusoes['A3'].font = Font(size=14, bold=True)
ws_conclusoes.merge_cells('A3:G3')

conclusoes = [
    "1. Avanços Significativos: A Rede de Atenção às Urgências na Bahia apresentou melhorias substanciais no período de 2013 a 2023.",
    "2. Persistência de Disparidades: Apesar dos avanços, persistem disparidades regionais importantes.",
    "3. Relação Estrutura-Resultado: Confirmou-se a forte correlação entre a disponibilidade de recursos e os desfechos clínicos.",
    "4. Impacto dos Processos: A melhoria nos processos assistenciais mostrou-se determinante para a redução da mortalidade.",
    "5. Integração com Atenção Básica: Regiões com maior cobertura de atenção básica apresentaram menor número de internações evitáveis."
]

for r_idx, texto in enumerate(conclusoes, 5):
    ws_conclusoes.cell(row=r_idx, column=1, value=texto)
    ws_conclusoes.merge_cells(f'A{r_idx}:G{r_idx}')

# Adicionar recomendações
ws_conclusoes['A11'] = "Recomendações"
ws_conclusoes['A11'].font = Font(size=14, bold=True)
ws_conclusoes.merge_cells('A11:G11')

recomendacoes = [
    ["Dimensão", "Recomendação", "Prazo", "Responsáveis"],
    ["Estrutura", "Ampliar a oferta de leitos de UTI nas regiões com menor disponibilidade.", "Médio prazo", "Secretaria Estadual de Saúde, Ministério da Saúde"],
    ["Estrutura", "Expandir a cobertura do SAMU 192 para atingir 100% da população do estado.", "Curto prazo", "Secretaria Estadual de Saúde, Secretarias Municipais"],
    ["Processo", "Implementar programas de educação continuada para adesão a protocolos clínicos.", "Curto prazo", "Secretaria Estadual de Saúde, Hospitais"],
    ["Processo", "Fortalecer os mecanismos de regulação, com centralização e informatização.", "Médio prazo", "Secretaria Estadual de Saúde"],
    ["Resultado", "Implementar sistema de monitoramento contínuo dos indicadores de resultado.", "Curto prazo", "Secretaria Estadual de Saúde"],
    ["Integração", "Fortalecer a integração entre a atenção básica e os serviços de urgência.", "Médio prazo", "Secretaria Estadual de Saúde, Secretarias Municipais"],
    ["Financiamento", "Revisar os mecanismos de financiamento, com alocação baseada em necessidades.", "Longo prazo", "Ministério da Saúde, Secretaria Estadual de Saúde"]
]

for r_idx, row in enumerate(recomendacoes, 13):
    for c_idx, value in enumerate(row, 1):
        ws_conclusoes.cell(row=r_idx, column=c_idx, value=value)
        if c_idx == 2:  # Coluna da recomendação
            ws_conclusoes.column_dimensions['B'].width = 50

formatar_cabecalhos(ws_conclusoes, 13)

# Ajustar larguras das colunas
for ws in [ws_resumo, ws_estrutura, ws_processo, ws_resultado, ws_conclusoes]:
    for col in ws.columns:
        max_length = 0
        column = col[0].column_letter
        for cell in col:
            if cell.value:
                try:
                    if len(str(cell.value)) > max_length:
                        max_length = len(str(cell.value))
                except:
                    pass
        adjusted_width = (max_length + 2)
        ws.column_dimensions[column].width = adjusted_width

# Salvar o arquivo Excel
excel_path = 'relatorio_iam_bahia.xlsx'
wb.save(excel_path)
print(f"Arquivo Excel salvo em: {excel_path}")
```

## Conclusão

Este notebook apresentou um resumo executivo da avaliação da efetividade da Rede de Atenção às Urgências no Estado da Bahia durante o período de 2013 a 2023. Os dados foram processados e visualizados utilizando o SUS Data Explorer, e os resultados foram exportados para formatos apresentáveis (Excel).

Os resultados mostram avanços significativos na estrutura, processos e resultados da rede, com destaque para a expansão da cobertura do SAMU, aumento do número de leitos de UTI, redução do tempo de resposta e diminuição da mortalidade por causas sensíveis à atenção às urgências. No entanto, persistem disparidades regionais importantes, que devem ser abordadas por meio de políticas públicas específicas.
